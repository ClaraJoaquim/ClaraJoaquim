```python
import os

from google.colab import userdata

os.environ['GOOGLE_API_KEY'] = userdata.get('GOOGLE_API_KEY')
```

```python
from google import genai

client = genai.Client();
```

```python
for model in client.models.list():
  print(model.name)
```

```python
modelo = "gemini-2.0-flash"

resposta = client.models.generate_content(model = modelo,
                                          contents = "Quem é a empresa por trás dos modelos do Gemini?")
```

```python
print(resposta);
```

```python
resposta
```

```python
resposta.text
```

```python
chat = client.chats.create(model = modelo);

resposta = chat.send_message("Oi tudo bem?");

resposta.text
```

```python
resposta = chat.send_message("Você é um assistente virtual e sempre responde de forma sucinta. O que é inteligência Artificial?");

resposta.text
```

```python
from google.genai import types

chat_config = types.GenerateContentConfig(
    system_instruction= "Você é um assistente virtual e sempre responde de forma sucinta.",

)

chat = client.chats.create(model = modelo, config = chat_config);
```

```python
resposta = chat.send_message("O que é ponto-cruz?");

resposta.text
```

```python
chat.get_history()
```

```python
prompt = input("Esperando prompt: ")

while (prompt != "fim") :
  resposta = chat.send_message(prompt)
  print("Resposta", resposta.text)
  print("\n")
  prompt = input("Esperando prompt: ")
```

```python
chat_config2 = types.GenerateContentConfig(
    system_instruction= "Você é um assistente virtual e sempre responde com muitos emojis, simulando uma amizade de anos com a pessoa que esta realizando as perguntas.",

)

chat_2 = client.chats.create(model = modelo, config = chat_config2);
```
